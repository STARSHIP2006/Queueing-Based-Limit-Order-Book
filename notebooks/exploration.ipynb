# LOB model exploration

This notebook is a starting point for inspecting simulated event paths, queue dynamics, and likelihood calculations. The reproducible experiment scripts live in `experiments/`.


```python
import sys
from pathlib import Path
import matplotlib.pyplot as plt

ROOT = Path.cwd()
if (ROOT / 'src').exists() is False and (ROOT.parent / 'src').exists():
    ROOT = ROOT.parent
if str(ROOT) not in sys.path:
    sys.path.insert(0, str(ROOT))

from src.simulator import LOBState, StructuralParameters, simulate_lob
from src.likelihood import log_likelihood
```

```python
params = StructuralParameters()
initial_state = LOBState(Qb=5, Qa=5, Pb=99, Pa=101)
event_log = simulate_lob(500.0, params, m=5.0, initial_state=initial_state, seed=2026)
event_log.head()
```

```python
fig, ax = plt.subplots(figsize=(10, 4))
ax.plot(event_log['t'], event_log['Qb_after'], label='Bid queue', linewidth=0.8)
ax.plot(event_log['t'], event_log['Qa_after'], label='Ask queue', linewidth=0.8)
ax.set(xlabel='Time (seconds)', ylabel='Queue size')
ax.legend()
plt.show()
```
